```python
import os
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

import os
```

```python
load_dotenv(override=True)  # Load environment variables from .env file
```

```python
llm = ChatOpenAI(
    model=os.getenv("CHAT_MODEL_DEPLOYMENT_NAME"),
    base_url=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY")
)
```

```python
template = """
You are an expert travel guide. 

Provide a 2-day itinerary for a trip to {destination}.
The traveler is highly interested in {interest}.

Format your response with clear headings for Day 1 and Day 2, along with top recommendations for food and activities.
"""

prompt = PromptTemplate(
    input_variables=["destination", "interest"],
    template=template
)
```

```python

# Create the Chain
# Using the LangChain Expression Language (LCEL), we chain the prompt and the LLM together using the `|` operator.
chain = prompt | llm
```

```python
# Run the Application
# We pass the input variables as a dictionary to the .invoke() method
inputs = {
    "destination": "Kyoto, Japan",
    "interest": "ancient history and traditional tea ceremonies"
}

print(f"Generating itinerary for {inputs['destination']} based on {inputs['interest']}...\n")
response = chain.invoke(inputs)
```

```python
# Print the output
# The response from a ChatModel is a ChatMessage object; .content extracts the text.
print(response.content)
```
