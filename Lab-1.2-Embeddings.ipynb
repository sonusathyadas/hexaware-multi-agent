```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)  # Load environment variables from .env file
```

```python
embeddings = OpenAIEmbeddings(
    model=os.getenv("EMBEDDING_MODEL_DEPLOYMENT_NAME"),
    base_url=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY")
)
```

```python
input_text = "The meaning of life is 42"
vector = embeddings.embed_query(input_text)
print(f"Last 3 dimensions of the vector: {vector[-3:]}")
print(f"Total dimensions: {len(vector)}")
```

```python
docs = [
    "Welcome to the world of embeddings.",
    "Embeddings are a way to represent text as vectors.",
    "They are useful for various NLP tasks.",
]

vectors = embeddings.embed_documents(docs)
print(f"Total documents embedded: {len(vectors)}")
print(f"Dimensions of the first document vector: {len(vectors[0])}")
print(f"Dimensions of the second document vector: {len(vectors[1])}")
print(f"Dimensions of the third document vector: {len(vectors[2])}")
```
